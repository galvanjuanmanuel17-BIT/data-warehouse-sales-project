```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files
uploaded = files.upload()

df = pd.read_excel('online_retail_with_reg_.xlsx')
```

```python
df.head()
df.info()
df.describe()
```

```python
#Limpieza

# Eliminar nulos clave
df=df[df['CustomerID'].notna()]

# Eliminar precios inválidos
df=df[df['UnitPrice']> 0]

# Separar devoluciones
df_returns = df[df['Quantity']< 0]
df = df[df['Quantity'] > 0]

df = df.copy()
```

```python
#Renombrar columnas
df.rename(columns={
    'InvoiceNo': 'Invoice_ID',
    'StockCode': 'Product_ID',
    'CustomerID': 'Customer_ID'
}, inplace=True)

#Ajustar tipo de dato
df['Customer_ID'] = df['Customer_ID'].astype(int)

#Crear métricas
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
#Tiempo
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Quarter'] = df['InvoiceDate'].dt.to_period('Q').astype(str)
```

```python
#Ordenar columnas
df = df[[
    'Invoice_ID',
    'Product_ID',
    'Description',
    'Quantity',
    'UnitPrice',
    'Revenue',
    'Customer_ID',
    'Country',
    'Region',
    'Categoria',
    'InvoiceDate',
    'Year',
    'Month',
    'Quarter'
]]
```

```python
df.to_csv('online_retail_clean.csv', index=False)
```

```python
file_name = "Online Retail.xlsx"
```

```python
df = pd.read_excel(file_name)
```

```python
country_to_region = {
    "United Kingdom": "Europe",
    "France": "Europe",
    "Germany": "Europe",
    "Netherlands": "Europe",
    "Belgium": "Europe",
    "Spain": "Europe",
    "Portugal": "Europe",
    "Italy": "Europe",
    "Switzerland": "Europe",
    "Austria": "Europe",
    "Norway": "Europe",
    "Sweden": "Europe",
    "Denmark": "Europe",
    "Finland": "Europe",
    "EIRE": "Europe",
    "Channel Islands": "Europe",
    "Poland": "Europe",
    "Japan": "Asia",
    "Singapore": "Asia",
    "Hong Kong": "Asia",
    "Israel": "Asia",
    "Lebanon": "Asia",
    "United Arab Emirates": "Asia",
    "Saudi Arabia": "Asia",
    "Australia": "Oceania",
    "New Zealand": "Oceania",
    "USA": "North America",
    "Canada": "North America",
    "Brazil": "South America"
}
```

```python
df["Region"] = df["Country"].map(country_to_region)
```

```python
df["Region"] = df["Region"].fillna("Other")
```

```python
output_file = "online_retail_with_region.xlsx"
df.to_excel(output_file, index=False)

print("Archivo creado:", output_file)
```

```python
df["Country"] = df["Country"].replace("EIRE", "Ireland")
```

```python
def clasificar_producto(desc):

    desc = str(desc).upper()

    if "CHRISTMAS" in desc or "TREE" in desc or "STAR" in desc:
        return "Navidad y temporada"

    elif "LANTERN" in desc or "CANDLE" in desc or "ORNAMENT" in desc:
        return "Decoración del hogar"

    elif "MUG" in desc or "PLATE" in desc or "BOWL" in desc or "CUP" in desc:
        return "Cocina y vajilla"

    elif "BAG" in desc or "BOX" in desc or "PAPER" in desc:
        return "Papelería y embalaje"

    elif "TOY" in desc or "DOLL" in desc:
        return "Juguetes"

    elif "LIGHT" in desc or "LAMP" in desc:
        return "Iluminación"

    else:
        return "Otros"


df["Categoria"] = df["Description"].apply(clasificar_producto)
```

```python
output_file = "online_retail_with_region.xlsx"
```

```python
df.to_excel(output_file, index=False)
```

```python
files.download(output_file)
```
